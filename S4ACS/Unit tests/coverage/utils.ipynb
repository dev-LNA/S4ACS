```python
# CREATE A LIST OF THE METHODS OF EACH S4ACS CLASS

import os

import pandas as pd

remove_functions = [
    "Define Test.vi",
    "Assert.vi",
    "Application Directory.vi",
    "Space Constant.vi",
]
tested_functions = pd.read_csv("tested_functions.csv")
tested_functions = tested_functions[
    tested_functions["node names"].str.contains(".vi", na=False)
]
tested_functions = tested_functions[
    ~tested_functions["node names"].isin(remove_functions)
]
```

```python
# CREATE A LIST OF THE METHODS OF EACH S4ACS CLASS

import os

import pandas as pd

all_functions = {"Class": [], "method": [], "covered": []}
subdirs = [d for d in os.listdir("..\\..") if os.path.isdir(os.path.join("..\\..", d))]
for folder_name in [
    "Unit tests",
    "communication-manager",
    "Comm VIs",
    "General VIs",
    "ACS files",
]:
    subdirs.remove(folder_name)

for subdir in subdirs:
    path = os.path.join("..\\..", subdir)
    class_methods = os.listdir(path)
    for class_method in class_methods:
        if ".lvclass" in class_method:
            continue
        all_functions["Class"].append(subdir)
        all_functions["method"].append(class_method)
        all_functions["covered"].append("")

all_functions = pd.DataFrame.from_dict(all_functions)
for idx, row in all_functions.iterrows():
    if row["method"] in tested_functions["node names"].values:
        all_functions.at[idx, "covered"] = "x"

counts = all_functions["covered"].str.contains("x", regex=False, na=False).sum()
all_functions["Percent covered"] = ""
all_functions.at[0, "Percent covered"] = f"{counts / all_functions.shape[0] * 100:.2f}"
all_functions.to_csv("test_coverage.csv", index=False, float_format="%.2f")
```
