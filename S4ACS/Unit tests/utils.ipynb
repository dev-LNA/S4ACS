```python
# CREATE A LIST OF THE METHODS OF EACH S4ACS CLASS

import os

import pandas as pd

all_methods = {"Class": [], "method": []}
subdirs = [d for d in os.listdir("..") if os.path.isdir(os.path.join("..", d))]
for folder_name in [
    "Unit tests",
    "communication-manager",
    "Comm VIs",
    "Events log file",
    "Filter_Wheel",
    "Error log",
    "General VIs",
    "Header content",
    "Keywords Log File",
    "S4ACS files",
    "Simulated_Header",
]:
    subdirs.remove(folder_name)

for subdir in subdirs:
    path = os.path.join("..", subdir)
    class_methods = os.listdir(path)
    for class_method in class_methods:
        if ".lvclass" in class_method:
            continue
        all_methods["Class"].append(subdir)
        all_methods["method"].append(class_method)

df = pd.DataFrame.from_dict(all_methods)
df.to_csv("test_coverage_new.csv", index=False)
```
